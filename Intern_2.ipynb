```python
s = input("Enter the sentence: ")
s = s.lower()
v = 0
for ch in s:
  if ch in "aeiou":
    v += 1
print(v)
```

Q7. Count Consonants in a String


```python
s = input("Enter the sentence: ")
s = s.lower()
c = 0
for ch in s:
  if ch not in "aeiou":
    c += 1
print(c)
```

Q10. Find All Characters Appearing More Than Once


```python
s = input("Enter the word: ")
di={}
for ch in s:
  di[ch] = di.get(ch,0) + 1
li=[]
for key in di.keys():
  if di[key] > 1:
    li.append(key)
print(li)
```

Q14. Create a Dictionary of Numbers and Their Squares


```python
n = int(input())
li=[]
for i in range(n):
  i = int(input("Enter the number: "))
  li.append(i)

di={}
for ch in li:
  di[ch]= di.get(ch,ch * ch)
print(di)
```

Q15. Create a Dictionary of Numbers and Their Even/Odd Status


```python
n = int(input())
li=[]
for i in range(n):
  i = int(input("Enter the number: "))
  li.append(i)

di={}
for ch in li:
  di[ch]= di.get(ch,'Even' if ch % 2 == 0 else "Odd")
print(di)
```

Q16. Find the Most Frequent Word in a Sentence


```python
s = input("Enter a sentence: ")
s = s.split()
di={}
for ch in s:
  di[ch]= di.get(ch,0) + 1
maxi = max(di.values())
for word in s:
  if di[word] == maxi:
    print(word)
    break;
```

Q18. Find the First Word That Appears Only Once


```python
s = input("Enter a sentence: ")
s = s.split()
di={}
for ch in s:
  di[ch]= di.get(ch,0) + 1

for word in s:
  if di[word] == 1:
    print(word)
    break;
```

Q19. Count Words Whose Length Is Greater Than 4


```python
s = input("Enter a sentence: ")
s = s.split()

print(sum(1 for word in s if len(word) > 4))
```

Q20. Find the Longest Word That Appears Only Once


```python
s = input("Enter a sentence: ")
s = s.split()
di={}
for ch in s:
  di[ch]= di.get(ch,0) + 1
unique = [word for word in s if di[word] == 1]

print(max(unique,key = len))
```
